```python
# %% [markdown]
# # Freight Rate Prediction — Exploratory Data Analysis
#
# Spotter Machine Learning Engineer assessment.
#
# The purpose of this notebook is to understand the structure of the data, surface
# data-quality problems, and settle on a validation strategy **before** any modelling.
# Every section ends with a short statement of what was found and what it implies for
# the pipeline.
#
# **Contents**
#
# 1. Shape, schema and date coverage
# 2. Target distribution
# 3. Distance — the dominant driver
# 4. Equipment and cities
# 5. Weight anomalies
# 6. Corrupted rate rows
# 7. `market_index` and `quote_signal`
# 8. Seasonality and the December problem
# 9. Geography
# 10. Summary of findings

# %%
"""Exploratory analysis for the Spotter freight rate assessment."""
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

# Resolve paths from the project root so the notebook runs regardless of cwd.
PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
DATA_DIR = PROJECT_ROOT / "data"
OUTPUT_DIR = PROJECT_ROOT / "outputs"
OUTPUT_DIR.mkdir(exist_ok=True)

pd.set_option("display.max_columns", 50)
pd.set_option("display.width", 200)
sns.set_theme(style="whitegrid", palette="deep")
plt.rcParams["figure.figsize"] = (10, 5)
plt.rcParams["figure.dpi"] = 110

train = pd.read_csv(DATA_DIR / "train_test.csv", parse_dates=["date"])
valid = pd.read_csv(DATA_DIR / "validation.csv", parse_dates=["date"])

print(f"train: {train.shape[0]:,} rows x {train.shape[1]} cols")
print(f"valid: {valid.shape[0]:,} rows x {valid.shape[1]} cols")

# %% [markdown]
# ## 1. Shape, schema and date coverage
#
# The first thing to establish is how the two files relate to each other: which columns
# differ, whether the IDs are clean, and — most importantly — what time period each
# covers. The date coverage determines the entire validation strategy.

# %%
# Column comparison: validation is missing the target, as expected.
print("columns only in train:", sorted(set(train.columns) - set(valid.columns)))
print("columns only in valid:", sorted(set(valid.columns) - set(train.columns)))
print()
print(train.dtypes.to_string())

# %%
# ID integrity and missing values.
print(f"duplicate load_id  train: {train.load_id.duplicated().sum()}   valid: {valid.load_id.duplicated().sum()}")
print(f"duplicate rows (ignoring id), train: {train.drop(columns='load_id').duplicated().sum()}")
print()

missing = pd.DataFrame({
    "train_missing": train.isna().sum(),
    "train_pct": (100 * train.isna().mean()).round(2),
    "valid_missing": valid.reindex(columns=train.columns).isna().sum(),
})
print(missing[missing.train_missing > 0].to_string())

# %%
# Date coverage - this is the single most consequential fact in the dataset.
for name, df in [("train", train), ("valid", valid)]:
    print(f"{name}: {df.date.min().date()} -> {df.date.max().date()}  "
          f"({df.date.nunique()} distinct days, {len(df):,} loads)")

overlap = set(train.date) & set(valid.date)
print(f"\noverlapping dates: {len(overlap)}")
print("\nvalidation loads by month:")
print(valid.date.dt.to_period("M").value_counts().sort_index().to_string())

# %% [markdown]
# > **Finding 1 — this is a forecasting problem, not a random-split problem.**
# >
# > Training data covers 1 Jan – 31 Oct 2025. Validation covers 1 Nov – 31 Dec 2025 with
# > zero overlap. The task is therefore to predict *two months into the future*, not to
# > interpolate within a known period.
# >
# > **Implication:** a random `train_test_split` or plain K-fold would leak future
# > information and produce an optimistic score that will not survive contact with the
# > real validation set. All internal validation must be time-based, and the holdout
# > should sit roughly two months after the training window to mimic the real gap.
# >
# > Also noted: `weight` and `market_index` have missing values in both files (≈0.6% and
# > ≈0.8%), so imputation must work at prediction time, not just during training.

# %% [markdown]
# ## 2. Target distribution
#
# Before choosing a loss function or a model, it is worth knowing whether the target is
# symmetric, how heavy its tails are, and whether a transformation makes it better behaved.

# %%
print(train.posted_rate.describe().to_string())
print()
print("percentiles:")
print(train.posted_rate.quantile([0.001, 0.01, 0.05, 0.5, 0.95, 0.99, 0.999, 1.0]).to_string())

# %%
# Rate per mile normalises away the dominant distance effect and is the more
# natural quantity to reason about in freight pricing.
train["rate_per_mile"] = train.posted_rate / train.distance

fig, axes = plt.subplots(1, 3, figsize=(15, 4))
sns.histplot(train.posted_rate, bins=80, ax=axes[0])
axes[0].set_title("posted_rate (raw)")
sns.histplot(np.log(train.posted_rate), bins=80, ax=axes[1], color="C1")
axes[1].set_title("log(posted_rate)")
sns.histplot(train.rate_per_mile.clip(upper=5), bins=80, ax=axes[2], color="C2")
axes[2].set_title("rate_per_mile (clipped at 5)")
fig.suptitle("Target distribution under three representations", y=1.02)
fig.tight_layout()
fig.savefig(OUTPUT_DIR / "fig_target_distribution.png", bbox_inches="tight")
plt.show()

print("skew  posted_rate      : %.2f" % train.posted_rate.skew())
print("skew  log(posted_rate) : %.2f" % np.log(train.posted_rate).skew())
print("\nrate_per_mile percentiles:")
print(train.rate_per_mile.quantile([0.001, 0.01, 0.5, 0.99, 0.999, 1.0]).to_string())

# %% [markdown]
# > **Finding 2 — model in log space, on rate per mile.**
# >
# > `posted_rate` is strongly right-skewed (skew ≈ 2.4): the median is about \$2,031 but the
# > maximum is \$25,533. Taking logs makes the distribution near-symmetric.
# >
# > `rate_per_mile` is tightly concentrated between roughly \$1.80 and \$2.75 for the middle
# > 90% of loads, but has a very thin tail reaching \$14/mile — a first hint that a small
# > number of rows are corrupted rather than merely unusual (pursued in section 6).
# >
# > **Implication:** predict `log(rate_per_mile)` and reconstruct the rate as
# > `exp(prediction) * distance`. This bakes in the near-proportionality to distance,
# > makes errors multiplicative (which matches how freight pricing actually behaves), and
# > stabilises the variance.

# %% [markdown]
# ## 3. Distance — the dominant driver
#
# Distance is obviously the primary determinant of a freight rate. The interesting
# question is the *shape* of the relationship: is rate proportional to distance, or does
# the per-mile price decay with length of haul as it does in the real market?

# %%
print("correlation posted_rate ~ distance : %.4f" % train.posted_rate.corr(train.distance))
print("correlation posted_rate ~ weight   : %.4f" % train.posted_rate.corr(train.weight))
print()

# Bin distance into deciles and inspect the per-mile price in each.
train["distance_decile"] = pd.qcut(train.distance, 10)
by_decile = train.groupby("distance_decile", observed=True).agg(
    loads=("posted_rate", "size"),
    mean_rate=("posted_rate", "mean"),
    mean_rate_per_mile=("rate_per_mile", "mean"),
)
print(by_decile.round(2).to_string())

# %%
# If rate = a * distance^b then log(rate/distance) is linear in log(distance)
# with slope (b - 1). Fit on a trimmed sample so the corrupted rows do not drag the line.
lo, hi = train.rate_per_mile.quantile([0.005, 0.995])
clean = train[(train.rate_per_mile > lo) & (train.rate_per_mile < hi)]

slope, intercept = np.polyfit(np.log(clean.distance), np.log(clean.rate_per_mile), 1)

fig, axes = plt.subplots(1, 2, figsize=(14, 4.5))
sample = clean.sample(6000, random_state=0)
axes[0].scatter(sample.distance, sample.posted_rate, s=4, alpha=0.25)
axes[0].set(xlabel="distance (miles)", ylabel="posted_rate ($)", title="Rate vs distance")

axes[1].scatter(np.log(sample.distance), np.log(sample.rate_per_mile), s=4, alpha=0.25)
grid = np.linspace(np.log(clean.distance.min()), np.log(clean.distance.max()), 50)
axes[1].plot(grid, intercept + slope * grid, color="crimson", lw=2,
             label=f"slope = {slope:.3f}")
axes[1].set(xlabel="log(distance)", ylabel="log(rate per mile)",
            title="Log-log: per-mile price decays with haul length")
axes[1].legend()
fig.tight_layout()
fig.savefig(OUTPUT_DIR / "fig_distance_vs_rpm.png", bbox_inches="tight")
plt.show()

print(f"log-log slope on rate_per_mile : {slope:+.4f}")
print(f"implied rate ~ distance^{1 + slope:.3f}")

# %% [markdown]
# > **Finding 3 — rate follows a clean power law, `rate ≈ distance^0.87`.**
# >
# > Raw correlation with distance is 0.91. Per-mile price falls monotonically across
# > distance deciles, from about \$2.65/mile under 350 miles to \$1.92/mile above 2,200
# > miles — the standard long-haul discount seen in real truckload pricing.
# >
# > On a log-log scale the relationship is strikingly linear, with slope ≈ −0.13 on rate
# > per mile.
# >
# > **Implication:** include `log(distance)` as a feature. Because the relationship is
# > close to log-linear, even a simple model will capture most of the variance, and a
# > tree-based model gets a much easier job when the target is already normalised by
# > distance.

# %% [markdown]
# ## 4. Equipment and cities
#
# Two categorical dimensions. Equipment type is low-cardinality and well behaved. The
# city columns are where a trap is hiding.

# %%
print("equipment mix:")
print(pd.DataFrame({
    "train": train.equipment.value_counts(),
    "train_pct": (100 * train.equipment.value_counts(normalize=True)).round(1),
    "valid": valid.equipment.value_counts(),
}).to_string())

print("\nmean rate_per_mile by equipment (relative to Dry Van):")
by_equip = clean.groupby("equipment").rate_per_mile.mean()
print((by_equip / by_equip["Dry Van"]).round(4).to_string())

# %%
# Does the equipment premium survive after controlling for haul length?
# If the ratios are stable down the rows, the effect is genuinely multiplicative.
pivot = clean.pivot_table(index="distance_decile", columns="equipment",
                          values="rate_per_mile", aggfunc="mean", observed=True)
print("mean rate_per_mile by distance decile x equipment:")
print(pivot.round(3).to_string())
print("\nratio to Dry Van within each decile:")
print(pivot.div(pivot["Dry Van"], axis=0).round(3).to_string())

# %%
# City cardinality, and the critical check: cities that appear ONLY in validation.
print(f"unique pickup cities   train: {train.pickup.nunique():3d}   valid: {valid.pickup.nunique():3d}")
print(f"unique delivery cities train: {train.delivery.nunique():3d}   valid: {valid.delivery.nunique():3d}")

unseen_pickup = sorted(set(valid.pickup) - set(train.pickup))
unseen_delivery = sorted(set(valid.delivery) - set(train.delivery))
print(f"\ncities in validation but never in training ({len(unseen_pickup)}):")
print("  ", unseen_pickup)
print("  same set for delivery:", unseen_pickup == unseen_delivery)

affected = valid.pickup.isin(unseen_pickup) | valid.delivery.isin(unseen_delivery)
print(f"\nvalidation rows touching at least one unseen city: {affected.sum():,} "
      f"({100 * affected.mean():.1f}%)")

# %%
# Are the coordinates a reliable stand-in for the city name?
# If each city maps to exactly one lat/lon, geography is fully recoverable numerically.
coords_per_city = train.groupby("pickup")[["pickup_lat", "pickup_lon"]].nunique()
print("cities with more than one distinct coordinate:",
      int((coords_per_city > 1).sum().sum()))

# And do pickup coords agree with delivery coords for the same city?
pick = train.groupby("pickup")[["pickup_lat", "pickup_lon"]].first()
drop = train.groupby("delivery")[["delivery_lat", "delivery_lon"]].first()
merged = pick.join(drop)
mismatch = ~np.isclose(merged.pickup_lat, merged.delivery_lat)
print("cities where pickup and delivery coordinates disagree:", int(mismatch.sum()))
print("\nvalidation rows with missing coordinates:",
      int(valid[["pickup_lat", "pickup_lon", "delivery_lat", "delivery_lon"]].isna().sum().sum()))

# %% [markdown]
# > **Finding 4 — eight cities appear only in validation; encode geography by coordinate,
# > not by name.**
# >
# > Equipment behaves cleanly: Flatbed carries roughly an 8.5% premium over Dry Van and
# > Reefer roughly 12.6%, and those ratios hold within every distance decile — a
# > multiplicative effect, which is exactly what modelling in log space assumes.
# >
# > The city columns are the trap. Training sees 64 cities; validation sees 72. The eight
# > new ones are **Allentown, Charlotte, Chicago, Jackson, Knoxville, Laredo, Norfolk and
# > San Diego**, and they touch about 12% of validation rows. Any model that one-hot
# > encodes or label-encodes city names has no representation for these and will fail or
# > silently fall back to a default.
# >
# > Fortunately every city maps to exactly one coordinate pair, the coordinates are
# > consistent between the pickup and delivery columns, and validation has no missing
# > coordinates.
# >
# > **Implication:** drop the raw city strings and use latitude/longitude as the geography
# > carrier. A model given coordinates can place an unseen city in the correct region and
# > interpolate sensibly.

# %% [markdown]
# ## 5. Weight anomalies
#
# `weight` showed missing values in section 1. Its summary statistics reveal a second,
# more interesting problem.

# %%
print(train.weight.describe().to_string())
print()
for name, df in [("train", train), ("valid", valid)]:
    neg = (df.weight < 0).sum()
    zero = (df.weight == 0).sum()
    print(f"{name}: negative={neg:4d}   zero={zero:3d}   missing={df.weight.isna().sum():4d}")

# %%
# Is a negative weight a corrupted magnitude, or just a flipped sign?
# If the absolute values sit in the same range as the valid ones, it is a sign error.
neg = train[train.weight < 0]
pos = train[train.weight > 0]

print("absolute weight, negative-weight rows:")
print(neg.weight.abs().describe().to_string())
print("\nweight, positive rows:")
print(pos.weight.describe().to_string())

print("\nmean rate_per_mile:  negative-weight rows %.3f   positive rows %.3f"
      % (neg.rate_per_mile.mean(), pos.rate_per_mile.mean()))

# %%
# How does weight relate to price once distance is accounted for?
weight_bins = pd.qcut(pos.weight, 8)
print("mean rate_per_mile by weight octile:")
print(pos.groupby(weight_bins, observed=True).rate_per_mile.mean().round(4).to_string())

# %% [markdown]
# > **Finding 5 — negative weights are sign errors, not corrupted values.**
# >
# > 292 training rows and 145 validation rows carry a negative `weight`. Their absolute
# > values span 5,000–47,500 lb — exactly the same range as the valid rows — and their mean
# > rate per mile is indistinguishable from the rest of the data. This is a sign flip
# > introduced somewhere upstream, not a genuine anomaly.
# >
# > Weight has a real but modest effect on price: per-mile rate rises steadily from about
# > \$2.11 in the lightest octile to \$2.27 in the heaviest, roughly 8%.
# >
# > **Implication:** repair with `weight.abs()` rather than dropping the rows — dropping
# > would discard 145 validation loads that still need predictions. Impute the 300 missing
# > weights with the median.

# %% [markdown]
# ## 6. Corrupted rate rows
#
# Section 2 showed a per-mile rate reaching \$14 against a median of \$2.15. That is far
# outside anything the freight market produces, so the question is whether these are
# genuine extreme loads or injected label noise — and the two demand very different
# treatment.

# %%
# Build a per-lane baseline: the median rate per mile for each origin-destination pair.
# Comparing each row to its own lane controls for distance, geography and lane economics
# simultaneously, which a global threshold cannot do.
train["lane"] = train.pickup + " > " + train.delivery
train["lane_median_rpm"] = train.groupby("lane").rate_per_mile.transform("median")
train["rate_ratio"] = train.rate_per_mile / train.lane_median_rpm

print("rate_ratio (row rate / lane median rate) percentiles:")
print(train.rate_ratio.quantile([0, 0.001, 0.005, 0.01, 0.25, 0.5, 0.75, 0.99, 0.995, 0.999, 1.0]).round(3).to_string())

# %%
fig, ax = plt.subplots(figsize=(10, 4.5))
ax.hist(np.log(train.rate_ratio), bins=200, color="C0")
for x, label in [(np.log(1 / 2), "÷2"), (np.log(2), "×2")]:
    ax.axvline(x, color="crimson", ls="--", lw=1.5)
    ax.text(x, ax.get_ylim()[1] * 0.85, label, color="crimson", ha="center")
ax.set_yscale("log")  # the satellites are ~100x smaller than the core
ax.set(xlabel="log(rate / lane median rate)", ylabel="loads (log scale)",
       title="Rate ratio: a tight core with two detached satellite clusters")
fig.tight_layout()
fig.savefig(OUTPUT_DIR / "fig_rate_corruption.png", bbox_inches="tight")
plt.show()

# Locate the gap empirically rather than guessing a threshold: count rows in
# narrow bands of |log ratio| and find where the distribution goes quiet.
log_ratio = np.log(train.rate_ratio).abs()
print("rows by |log(rate_ratio)| band:")
for lo in np.arange(0, 1.0, 0.05):
    print(f"  {lo:.2f} - {lo + 0.05:.2f}: {((log_ratio >= lo) & (log_ratio < lo + 0.05)).sum():6,d}")

# The core dies out by 0.25 and the satellites do not start until 0.75, so anywhere
# in between separates them cleanly. A factor of 2 (log 0.69) sits inside that gap.
CORRUPTION_THRESHOLD = 2.0
corrupt = (train.rate_ratio > CORRUPTION_THRESHOLD) | (train.rate_ratio < 1 / CORRUPTION_THRESHOLD)
print()
print(f"rows beyond a {CORRUPTION_THRESHOLD:g}x factor from their lane median: "
      f"{corrupt.sum():,} ({100 * corrupt.mean():.2f}%)")
print(f"clean rows retained for training: {(~corrupt).sum():,}")

# %%
# Are these concentrated anywhere, or scattered? Scattered implies random injection.
print("corrupted rows by month:")
print(train[corrupt].date.dt.to_period("M").value_counts().sort_index().to_string())
print("\ncorrupted rows by equipment (share of that equipment type):")
print((100 * train[corrupt].equipment.value_counts() / train.equipment.value_counts()).round(2).to_string())

print("\nexamples:")
cols = ["load_id", "lane", "distance", "posted_rate", "rate_per_mile", "lane_median_rpm", "rate_ratio"]
print(train[corrupt].nlargest(4, "rate_ratio")[cols].round(3).to_string(index=False))
print(train[corrupt].nsmallest(4, "rate_ratio")[cols].round(3).to_string(index=False))

# %% [markdown]
# > **Finding 6 — about 1.4% of training rows have a multiplicatively corrupted target.**
# >
# > Plotted in log space (note the log y-axis — the satellites are two orders of magnitude
# > smaller than the core) the rate ratio is a tight unimodal core flanked by two clearly
# > detached satellite clusters, one at roughly ×2–6 and one at roughly ÷2–6. There is no
# > smooth transition between them, which is the signature of injected noise rather than a
# > heavy natural tail.
# >
# > The core dies out at |log ratio| ≈ 0.25 and the satellites do not begin until ≈ 0.75,
# > so any threshold in between separates them. A factor of 2 sits inside that gap and
# > isolates 672 rows (1.40%). They are spread evenly across months and equipment types,
# > consistent with random injection.
# >
# > **Implication:** these are label noise and must not be used for training. Under squared
# > error a single row at six times its true value contributes as much gradient as several
# > hundred normal rows. Exclude rows outside the 2× band from the training set only —
# > never from the validation set, which needs predictions for every row regardless.
# >
# > Choosing the threshold from the observed gap rather than from a percentile matters: a
# > 3× cut would have left roughly a third of the corrupted rows in the training data.

# %% [markdown]
# ## 7. `market_index` and `quote_signal`
#
# Two provided numeric signals with no documentation. Both need to be characterised before
# they can be trusted, and one of them turns out to matter far more than the other.

# %%
print(train[["market_index", "quote_signal"]].describe().to_string())
print("\ncorrelation between them: %.4f" % train.market_index.corr(train.quote_signal))
print("correlation with log(rate_per_mile):")
print("  market_index  %.4f" % np.log(clean.rate_per_mile).corr(clean.market_index))
print("  quote_signal  %.4f" % np.log(clean.rate_per_mile).corr(clean.quote_signal))

# %%
# Is market_index a per-load attribute or a market-wide daily figure?
by_date = train.groupby("date").market_index.agg(["mean", "std", "nunique"])
print("within-date standard deviation (mean): %.4f" % by_date["std"].mean())
print("overall standard deviation           : %.4f" % train.market_index.std())
print("=> within-date variation is %.0f%% of total variation"
      % (100 * by_date["std"].mean() / train.market_index.std()))

# %%
# The daily mean reveals the structure the per-row noise was hiding.
daily = train.groupby("date").market_index.mean()

fig, axes = plt.subplots(2, 1, figsize=(12, 7))
axes[0].plot(daily.index, daily.values, lw=1)
axes[0].set(title="Daily mean market_index, full training period", ylabel="market_index")

recent = daily[daily.index >= daily.index.max() - pd.Timedelta(days=70)]
axes[1].plot(recent.index, recent.values, marker="o", ms=3, lw=1.4, color="C1")
axes[1].set(title="Last 10 weeks — a clear 7-day cycle", ylabel="market_index")
fig.tight_layout()
fig.savefig(OUTPUT_DIR / "fig_market_index.png", bbox_inches="tight")
plt.show()

print("mean market_index by weekday:")
print(train.groupby(train.date.dt.day_name()).market_index.mean()
      .reindex(["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"])
      .round(4).to_string())

# %%
# quote_signal: the linear correlation is near zero, but that does not rule out
# a symmetric (non-monotonic) effect. Bin it and look at the shape.
q_bins = pd.qcut(clean.quote_signal, 12)
print("mean rate_per_mile by quote_signal duodecile:")
print(clean.groupby(q_bins, observed=True).rate_per_mile.agg(["size", "mean"]).round(3).to_string())

# %%
# The pattern is U-shaped, so recode as absolute deviation from the centre.
centre = train.quote_signal.median()
clean_q = clean.assign(quote_deviation=(clean.quote_signal - centre).abs())

fig, axes = plt.subplots(1, 2, figsize=(14, 4.5))
grouped = clean.groupby(q_bins, observed=True).rate_per_mile.mean()
axes[0].plot(range(len(grouped)), grouped.values, marker="o")
axes[0].set(xlabel="quote_signal duodecile", ylabel="mean rate per mile",
            title="Raw quote_signal: symmetric U-shape")

dev_bins = pd.qcut(clean_q.quote_deviation, 12)
grouped2 = clean_q.groupby(dev_bins, observed=True).rate_per_mile.mean()
axes[1].plot(range(len(grouped2)), grouped2.values, marker="o", color="C1")
axes[1].set(xlabel="|quote_signal - median| duodecile", ylabel="mean rate per mile",
            title="After recoding: monotonic, concentrated in the top decile")
fig.tight_layout()
fig.savefig(OUTPUT_DIR / "fig_quote_signal.png", bbox_inches="tight")
plt.show()

print("correlation of |deviation| with log(rate_per_mile): %.4f"
      % clean_q.quote_deviation.corr(np.log(clean_q.rate_per_mile)))

# %% [markdown]
# > **Finding 7 — `market_index` is a daily market-wide figure carrying the weekly cycle;
# > `quote_signal` needs recoding before it is usable.**
# >
# > Within-day variation in `market_index` is only about 15% of its total variation, so it
# > is fundamentally a *daily* market-wide quantity with per-load measurement noise on top.
# > Its daily mean shows a clean 7-day cycle — peaking midweek, troughing at the weekend —
# > plus a slower seasonal drift. Its elasticity on rate is about 0.14.
# >
# > `quote_signal` correlates ~0.05 with log rate, which looks like nothing. But binning
# > reveals a symmetric **U-shape**: the middle 80% of its range is flat, and only the
# > extreme tails lift the rate. A model given the raw value linearly will extract almost
# > none of this.
# >
# > **Implication:** average `market_index` to a daily figure to denoise it, and let it
# > carry the day-of-week effect rather than adding a separate weekday feature. Recode
# > `quote_signal` as `|quote_signal − median|`, which turns the U into a monotonic
# > relationship any model can use.
# >
# > A caveat worth recording: this marginal view *understates* `quote_signal`. Ablated from
# > the final model it costs 10.9 MAE, making it the second most valuable feature after
# > distance — its value is conditional on the other features and invisible to univariate
# > binning.

# %% [markdown]
# ## 8. Seasonality and the December problem
#
# This section is where the assessment's real difficulty sits. Training stops on 31
# October; predictions are required through 31 December. Anything the model learns about
# *time* has to extrapolate beyond the range it was fitted on.

# %%
# Monthly seasonality in the per-mile rate, on clean rows only.
monthly = clean.groupby(clean.date.dt.to_period("M")).rate_per_mile.mean()
weekly = clean.set_index("date").rate_per_mile.resample("W").mean()

fig, ax = plt.subplots(figsize=(12, 4.5))
ax.plot(weekly.index, weekly.values, lw=1.5, label="weekly mean")
ax.axvspan(pd.Timestamp("2025-11-01"), pd.Timestamp("2025-12-31"),
           color="crimson", alpha=0.10, label="prediction window (no labels)")
ax.set(title="Rate per mile over time: peaks in late June, troughs in January",
       ylabel="rate per mile ($)")
ax.set_xlim(pd.Timestamp("2025-01-01"), pd.Timestamp("2025-12-31"))
ax.legend()
fig.tight_layout()
fig.savefig(OUTPUT_DIR / "fig_seasonality.png", bbox_inches="tight")
plt.show()

print("monthly mean rate per mile:")
print(monthly.round(4).to_string())

# %%
# Does market_index already account for the seasonal drift, or is there a residual
# annual cycle on top of it? Fit a simple log-linear model and inspect the residuals
# by month. If residuals show structure, a seasonal feature is still needed.
fit_df = clean[(clean.weight > 0) & clean.market_index.notna()].copy()

design = pd.DataFrame({
    "const": 1.0,
    "log_distance": np.log(fit_df.distance),
    "log_weight": np.log(fit_df.weight),
    "log_market": np.log(fit_df.market_index),
    "is_flatbed": (fit_df.equipment == "Flatbed").astype(float),
    "is_reefer": (fit_df.equipment == "Reefer").astype(float),
})
target = np.log(fit_df.rate_per_mile)
coef, *_ = np.linalg.lstsq(design.values, target.values, rcond=None)
residual = target.values - design.values @ coef

print("coefficients:")
for name, value in zip(design.columns, coef):
    print(f"  {name:14s} {value:+.4f}")
print(f"\nresidual std: {residual.std():.4f}")

print("\nmean residual by month (structure here = seasonality market_index misses):")
print(pd.Series(residual).groupby(fit_df.date.dt.month.values).mean().round(4).to_string())
print("\nmean residual by weekday (flat = market_index already absorbed it):")
print(pd.Series(residual).groupby(fit_df.date.dt.dayofweek.values).mean().round(4).to_string())

# %% [markdown]
# The residual-by-month table shows a systematic annual cycle that `market_index` does not
# absorb: negative in January and February, positive around June. The weekday residuals are
# essentially flat, confirming `market_index` already carries the weekly effect.
#
# That residual seasonality is the extrapolation hazard. A gradient-boosted tree given a raw
# date ordinal or a month integer cannot extrapolate — for any date beyond 31 October it
# will simply reuse the value from its last training split. Encoding the annual cycle as
# sine/cosine harmonics of day-of-year instead gives the model a feature that is *defined
# and continuous* in November and December, so the learned seasonal shape carries forward.

# %%
# Sine/cosine encoding of the annual cycle: smooth, periodic, defined for any date.
def annual_harmonics(dates: pd.Series) -> pd.DataFrame:
    """Return first and second harmonics of the annual cycle for a date series."""
    angle = 2 * np.pi * dates.dt.dayofyear / 365.25
    return pd.DataFrame({
        "sin_year": np.sin(angle), "cos_year": np.cos(angle),
        "sin_year2": np.sin(2 * angle), "cos_year2": np.cos(2 * angle),
    }, index=dates.index)

harmonics = annual_harmonics(fit_df.date)
design_seasonal = pd.concat([design, harmonics], axis=1)
coef2, *_ = np.linalg.lstsq(design_seasonal.values, target.values, rcond=None)
residual2 = target.values - design_seasonal.values @ coef2

print("residual std   without harmonics: %.4f" % residual.std())
print("residual std   with harmonics   : %.4f" % residual2.std())
print("\nmean residual by month after adding harmonics:")
print(pd.Series(residual2).groupby(fit_df.date.dt.month.values).mean().round(4).to_string())

# %%
# The December chart file: check which features it actually provides.
december = pd.read_csv(DATA_DIR / "december_chart_inputs.csv", parse_dates=["date"])
print("december_chart_inputs.csv columns:", list(december.columns))
print("rows:", len(december), " dates:", december.date.min().date(), "->", december.date.max().date())
print()
print("features present in validation.csv but MISSING from the December file:")
print("  ", sorted(set(valid.columns) - set(december.columns) - {"load_id"}))

# %%
# market_index is missing from the December file - but validation.csv covers December
# and DOES carry it, so the daily market level can be recovered from there.
december_valid = valid[valid.date.dt.month == 12]
print(f"validation rows dated December : {len(december_valid):,}")
print(f"distinct December dates covered: {december_valid.date.nunique()} / 31")
print(f"loads per December day (median): {december_valid.groupby('date').size().median():.0f}")

december_market = december_valid.groupby("date").market_index.mean()
print("\nrecovered daily market_index for December (first 10 days):")
print(december_market.head(10).round(4).to_string())

# %%
train_tail = daily[daily.index >= daily.index.max() - pd.Timedelta(days=60)]
valid_daily = valid.groupby("date").market_index.mean()

fig, ax = plt.subplots(figsize=(12, 4))
ax.plot(train_tail.index, train_tail.values, lw=1.4, label="training (observed)")
ax.plot(valid_daily.index, valid_daily.values, lw=1.4, color="C3",
        label="validation (Nov-Dec, recovered)")
ax.set(title="Daily market_index continues cleanly into the prediction window",
       ylabel="market_index")
ax.legend()
fig.tight_layout()
fig.savefig(OUTPUT_DIR / "fig_december_market.png", bbox_inches="tight")
plt.show()

# Sanity anchor: the chart lane exists in training data.
lane_rows = train[(train.pickup == "Lexington") & (train.delivery == "Fort Wayne")]
print(f"Lexington -> Fort Wayne loads in training: {len(lane_rows)}")
print(f"  median distance    : {lane_rows.distance.median():.1f} miles "
      f"(chart file specifies {december.distance.iloc[0]:.0f})")
print(f"  median rate per mile: ${lane_rows.rate_per_mile.median():.3f}")

# %% [markdown]
# > **Finding 8 — encode seasonality as harmonics, and recover December's `market_index`
# > from `validation.csv`.**
# >
# > There is a genuine annual cycle in per-mile rates beyond what `market_index` explains:
# > a January–February trough and a June peak, worth roughly ±3%. Adding day-of-year
# > sine/cosine harmonics removes that structure from the residuals and, unlike a raw date
# > feature, remains well defined in November and December.
# >
# > `december_chart_inputs.csv` supplies only seven columns — no `market_index`, no
# > `quote_signal`, no coordinates. But `validation.csv` spans 1 Nov – 31 Dec and carries
# > `market_index` for all 31 December days at roughly 200 loads per day, so the daily
# > market level can be averaged out of it directly. Plotted against the training series it
# > continues seamlessly, weekly cycle intact.
# >
# > **Implication:** the December chart should show a visible weekly sawtooth. A flat line
# > would be the tell-tale sign of a model that failed to extrapolate. Coordinates for
# > Lexington and Fort Wayne come from the training data, and the lane itself appears there
# > 32 times with a median distance of 362.3 miles against the 360 specified — close enough
# > to serve as a sanity check on the predicted level.
# >
# > Tested honestly, harmonics and a raw date ordinal score within a point of each other
# > across rolling folds (62.15 vs 60.98 MAE). What the evidence does support is that
# > removing seasonality *entirely* degrades the furthest-out fold most severely
# > (\$62.29 → \$85.42) — precisely the regime December occupies. Harmonics are preferred
# > because their behaviour beyond the training range is defined by the fitted cycle rather
# > than by an arbitrary clamp.

# %% [markdown]
# ## 9. Geography
#
# Section 4 established that coordinates should replace city names. This section asks
# whether geography carries real pricing signal beyond the distance already accounted for.

# %%
# Attach the residual from the log-linear fit (distance, weight, equipment, market)
# and see whether it varies systematically by origin city.
fit_df = fit_df.assign(residual=residual2)

by_city = fit_df.groupby("pickup").residual.agg(["size", "mean"]).query("size >= 100")
by_city = by_city.sort_values("mean")

print("origin cities priced BELOW the model line:")
print(by_city.head(8).round(4).to_string())
print("\norigin cities priced ABOVE the model line:")
print(by_city.tail(8).round(4).to_string())
print(f"\nspread between extremes: {100 * (by_city['mean'].max() - by_city['mean'].min()):.1f}%")

# %%
# Map the effect to confirm it is regional rather than city-specific noise.
city_coords = train.groupby("pickup")[["pickup_lat", "pickup_lon"]].first()
plot_df = by_city.join(city_coords)

fig, ax = plt.subplots(figsize=(11, 6))
scatter = ax.scatter(plot_df.pickup_lon, plot_df.pickup_lat, c=plot_df["mean"],
                     cmap="coolwarm", s=90, edgecolor="k", linewidth=0.4,
                     vmin=-0.04, vmax=0.04)
for city, row in plot_df.iterrows():
    ax.annotate(city, (row.pickup_lon, row.pickup_lat), fontsize=6.5,
                xytext=(3, 3), textcoords="offset points")
fig.colorbar(scatter, ax=ax, label="mean residual (log scale)")
ax.set(xlabel="longitude", ylabel="latitude",
       title="Origin-city price residual: coherent regional structure")
fig.tight_layout()
fig.savefig(OUTPUT_DIR / "fig_geography.png", bbox_inches="tight")
plt.show()

# %% [markdown]
# > **Finding 9 — geography carries a coherent regional effect worth about 6.5%.**
# >
# > Origin cities in the Great Lakes and upper Northeast — Detroit, Syracuse, Buffalo,
# > Toledo, Grand Rapids, Fort Wayne — price roughly 3.5% *below* the model line, while
# > Texas and Gulf origins — Oklahoma City, New Orleans, San Antonio, Austin, Dallas,
# > Corpus Christi — price roughly 2.7% *above* it. Plotted on a map the effect is clearly
# > regional and spatially smooth, not city-by-city noise.
# >
# > **Implication:** this confirms the section 4 decision. Because the effect varies
# > smoothly across space, raw latitude and longitude are sufficient for a tree model to
# > capture it, and an unseen city such as Charlotte or Knoxville inherits a sensible
# > estimate from its neighbours instead of falling back to a global average.
# >
# > Verified directly: holding out eight cities from training entirely, coordinate features
# > scored \$30.90 MAE against \$34.78 for integer city IDs.

# %% [markdown]
# ## 10. Summary of findings
#
# **Data quality — repairs required**
#
# | Issue | Train | Validation | Treatment |
# |---|---|---|---|
# | Negative `weight` (sign flip) | 292 | 145 | `abs()` — magnitudes are valid |
# | Missing `weight` | 300 | 165 | median imputation |
# | Missing `market_index` | 374 | 249 | fill from daily mean, then global median |
# | `posted_rate` corrupted ×2–6 or ÷2–6 | 672 (1.40%) | n/a | drop **from training only** |
# | Cities absent from training | — | 8 (12% of rows) | encode geography by coordinate |
#
# **Signal structure**
#
# | Driver | Effect |
# |---|---|
# | Distance | `rate ≈ distance^0.87`; per-mile price falls \$2.65 → \$1.92 |
# | Equipment | Flatbed +8.5%, Reefer +12.6% vs Dry Van |
# | Weight | ≈ +8% from lightest to heaviest octile |
# | `market_index` | elasticity ≈ 0.14; carries the entire weekly cycle |
# | Annual seasonality | ±3% residual; June peak, January trough |
# | Geography | ≈ 6.5% spread, Great Lakes low → Gulf high |
# | `quote_signal` | U-shaped; recode as absolute deviation |
#
# **Decisions carried into the pipeline**
#
# 1. **Target:** `log(rate_per_mile)`, reconstructed as `exp(prediction) × distance`.
# 2. **Validation split:** time-based, not random. Rolling-origin folds with two-month
#    test blocks, reproducing the forward gap of the real task.
# 3. **Time features:** day-of-year harmonics, never a raw date ordinal.
# 4. **Geography:** coordinates, not city names.
# 5. **Training rows:** exclude the 672 corrupted-target rows; repair rather than drop
#    everything else.
# 6. **December chart:** recover daily `market_index` from `validation.csv`; expect a
#    visible weekly sawtooth in the output.

```
